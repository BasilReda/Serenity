## Imports

```python
import os
import urllib.request

import fasttext
from huggingface_hub import hf_hub_download
```

## Helper Functions


```python
def download_model(url: str, save_path: str):
    os.makedirs(os.path.dirname(save_path), exist_ok=True)

    if not os.path.exists(save_path):
        print("Downloading model...")
        urllib.request.urlretrieve(url, save_path)
        print(f"Download completed: '{save_path}'")
    else:
        print(f"Model already exists: '{save_path}'")
```

```python
def predict_language(model, text: str):
    labels, scores = model.predict(text)

    language = labels[0].replace("__label__", "")

    print(
        f"Text: {text}\n"
        f"Language: {language}\n"
        f"Confidence: {scores[0]:.4f}\n"
    )
```

## Sample Texts

```python
samples = [
    "Hello, world!",
    "مرحبا كيف حالك",
    "নমস্কার",
    "こんにちは世界",
]
```

---

## 1. FastText lid.176.ftz

> Compressed model (~1 MB)


### Preparing model

```python
MODEL_URL = "https://dl.fbaipublicfiles.com/fasttext/supervised-models/lid.176.ftz"
MODEL_PATH = f"models/{MODEL_URL.split("/")[-1]}"

download_model(MODEL_URL, MODEL_PATH)
```

```python
ftz_model = fasttext.load_model(MODEL_PATH)
```

### Predictions

```python
for text in samples:
    predict_language(ftz_model, text)
```

---

## 2. FastText lid.176.bin


> Full model (~126 MB)

### Preparing model

```python
MODEL_URL = "https://dl.fbaipublicfiles.com/fasttext/supervised-models/lid.176.bin"
MODEL_PATH = f"models/{MODEL_URL.split("/")[-1]}"

download_model(MODEL_URL, MODEL_PATH)
```

```python
bin_model = fasttext.load_model(MODEL_PATH)
```

### Predictions

```python
for text in samples:
    predict_language(bin_model, text)
```

## 3. Meta FastText Language Identification

Hugging Face:
https://huggingface.co/facebook/fasttext-language-identification


> Hugging Face model (~1.18 GB)

### Preparing model

```python
hf_model_path = hf_hub_download(
    repo_id="facebook/fasttext-language-identification",
    filename="model.bin"
    )

hf_model = fasttext.load_model(hf_model_path)
```

### Predictions

```python
for text in samples:
    predict_language(hf_model, text)
```
